## 2. Statistical Aggregation & Guardrail Metrics
Here we group participants based on their daily viewing time (`< 60 mins` vs `>= 60 mins`). We calculate the mean and median procrastination scores. 

To quantify uncertainty, we use **Bootstrap resampling (2000 iterations, random seed = 42)** to calculate the 95% Confidence Intervals (CI). Additionally, we track a Guardrail Metric: the percentage of participants with a high procrastination score (score > 50).

```python
def bootstrap_ci(data, n_boot=2000, ci=95):
    boot_means = [np.mean(np.random.choice(data, size=len(data), replace=True)) for _ in range(n_boot)]
    lower = np.percentile(boot_means, (100 - ci) / 2)
    upper = np.percentile(boot_means, 100 - (100 - ci) / 2)
    return lower, upper

# Делим на группы по времени просмотра
df['viewing_category'] = np.where(df['viewing_minutes'] < 60, '< 60 mins', '>= 60 mins')

summary_data = []
for group_name, group_data in df.groupby('viewing_category'):
    scores = group_data['procrastination_score']
    mean_val = np.mean(scores)
    median_val = np.median(scores)
    ci_low, ci_high = bootstrap_ci(scores)
    error_rate = np.mean(scores > 50) * 100 # Guardrail metric
    
    summary_data.append({
        'Group': group_name, 'Count': len(scores), 'Mean_Score': mean_val,
        'Median_Score': median_val, 'CI_Lower': ci_low, 'CI_Upper': ci_high,
        'High_Procrastination_Rate_%': error_rate
    })

summary_df = pd.DataFrame(summary_data)

# Сохраняем результат
output_path = ROOT / "data/processed/metrics_summary.csv"
output_path.parent.mkdir(parents=True, exist_ok=True)
summary_df.to_csv(output_path, index=False)

display(summary_df.round(2))
print("Файл сохранен:", output_path)
```

## 3. Scientific Visualization
We generate two plots adhering to the 3 core rules of scientific visualization:
1. **Explicit Axes & Units:** All axes have descriptive titles with physical units (Minutes, Points).
2. **Uncertainty Quantification:** Figure 1 includes 95% CI error bars.
3. **Accessibility:** We use a colorblind-safe palette (`sns.color_palette("colorblind")`) and distinct markers. 

Both plots are exported as `.pdf` (vector) and `.png` (300 DPI raster) to the `reports/figures/` directory.

```python
fig_dir = ROOT / "reports/figures"
fig_dir.mkdir(parents=True, exist_ok=True)

# --- ГРАФИК 1 ---
fig1, ax1 = plt.subplots(figsize=(8, 6))
x = np.arange(len(summary_df))
y = summary_df['Mean_Score']
yerr = [y - summary_df['CI_Lower'], summary_df['CI_Upper'] - y]

ax1.bar(x, y, yerr=yerr, align='center', alpha=0.7, ecolor='black', capsize=10, 
        color=[palette[0], palette[1]], edgecolor='black')
ax1.set_xticks(x)
ax1.set_xticklabels(summary_df['Group'])
ax1.set_title("Effect of Viewing Time on Procrastination Score")
ax1.set_xlabel("Daily Viewing Time Category")
ax1.set_ylabel("Procrastination Score (Points)")
plt.tight_layout()

fig1.savefig(fig_dir / "fig1_procrastination_vs_viewing.pdf", bbox_inches='tight')
fig1.savefig(fig_dir / "fig1_procrastination_vs_viewing.png", dpi=300, bbox_inches='tight')
plt.show()

# --- ГРАФИК 2 ---
fig2, ax2 = plt.subplots(figsize=(8, 6))
sns.regplot(x='viewing_minutes', y='procrastination_score', data=df, 
            ax=ax2, color=palette[2], scatter_kws={'alpha':0.5, 'marker': 'o'})
ax2.set_title("Correlation between Viewing Minutes and Procrastination")
ax2.set_xlabel("Daily Viewing Time (Minutes)")
ax2.set_ylabel("Procrastination Score (Points)")
plt.tight_layout()

fig2.savefig(fig_dir / "fig2_viewing_vs_procrastination.pdf", bbox_inches='tight')
fig2.savefig(fig_dir / "fig2_viewing_vs_procrastination.png", dpi=300, bbox_inches='tight')
plt.show()

print("Графики успешно сохранены!")
```

## 4. Results & Discussion

Based on the analysis of real survey data, we can draw the following conclusions:

1. **Impact of Viewing Time:** Participants who watch short-form videos for **60 minutes or more per day** have a significantly higher mean procrastination score compared to those who watch less than 60 minutes.
2. **Guardrail Metric:** The rate of high procrastination (score > 50) is dramatically higher in the `>= 60 mins` group.
3. **Uncertainty:** The 95% Bootstrap Confidence Intervals do not overlap, suggesting that the difference between the two groups is statistically significant.

The scatter plot further confirms a positive correlation between daily viewing minutes and procrastination levels. However, this being observational data, correlation does not imply causation (e.g., people who are already prone to procrastination might simply watch more videos).
